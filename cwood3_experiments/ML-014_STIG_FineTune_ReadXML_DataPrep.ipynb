```python
###########################################
#- Data cleaning, way more code and defenses here...
###########################################
def cleanse_text(inc_text:str) -> str:

    resultant=inc_text.rstrip('\\r\\n').lstrip("b'")
    resultant=str(resultant).strip('\\n')
    resultant=str(resultant).strip('\n')
    ' '.join(resultant.split('\n'))

    tokens = word_tokenize(resultant)
    # convert to lower case
    tokens = [w.lower() for w in tokens]

    # remove punctuation from each word
    table = str.maketrans('', '', string.punctuation)
    stripped = [w.translate(table) for w in tokens]
    
    # remove remaining tokens that are not alphabetic
    words = [word for word in stripped if word.isalpha()]
    
    # filter out stop words
    stop_words = set(stopwords.words('english'))
    words = [w for w in words if not w in stop_words]

    # normalize words
    #porter = PorterStemmer()
    #stemmed = [porter.stem(word) for word in words]

    resultant = " ".join(words)
    
    return resultant

###########################################
#- STIG prompt builder
###########################################
def stig_prompt_builder(key_field: str, stig_reference: str, input_field: str) -> str:
    return f"What is the {key_field} for the Security Technical Implementation Guide (STIG) {stig_reference} regarding {input_field}"
```

```python
###########################################
#- Generic 
###########################################
#Example format for Google interactions with their models
#https://cloud.google.com/vertex-ai/generative-ai/docs/models/tune-text-models-supervised#sample-datasets
#https://github.com/GoogleCloudPlatform/vertex-ai-samples/blob/main/notebooks/official/generative_ai/tune_peft.ipynb

def create_perf_response_template (prompt_content: str, completion_content: str) -> str:
    data={
        "input": prompt_content,
        "output": completion_content,
    }
    json_string=json.dumps(data)
    return(json_string)

def create_pcpf_response_template (prompt_content: str, completion_content: str) -> str:
    data={
        "prompt": prompt_content,
        "completion": completion_content,
    }
    json_string=json.dumps(data)
    return(json_string)

###########################################
#- Chat GPT 3.5 Style
###########################################
#Example necessary for ChatGPT 3.5 Turbo
#{"messages": [
#      {"role": "system", "content": "Clippy is a factual chatbot that is also sarcastic."}, 
#      {"role": "user", "content": "Who discovered Antarctica?"}, 
#      {"role": "assistant", "content": "Some chaps named Fabian Gottlieb von Bellingshausen and Mikhail Lazarev, as if they don't teach that in every school!"}]}
def create_gpt35_response_template(system_content: str, user_content: str, assistant_content: str) -> str:
    data = {
        "messages": [
            {"role": "system", "content": system_content},
            {"role": "user", "content": user_content},
            {"role": "assistant", "content": assistant_content}
        ]
    }
    json_string=json.dumps(data)
    return(json_string)

def full_input(inc_model: str,
               the_name: str,
               rule_id: str,
               severity: str,
               title: str,
               description: str,
               fix_test:str , 
               check_test:str ) -> str:

    full_payload=", ".join(["ID: " + rule_id, "SEVERITY: " + severity, "TITLE:" + title, "DESCRIPTION: " + description, "FIX TEST: " + fix_test, "CHECK TEST: " + check_test])
    if inc_model=="perf":
        return(create_perf_response_template(rule_id, full_payload))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(rule_id, full_payload))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM, rule_id, full_payload))

def id_relation(inc_model: str,
                inc_name:  str, 
                inc_title: str, 
                inc_rule:  str) -> str:

    the_answer=stig_prompt_builder("id", inc_name, inc_title)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_rule}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_rule}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM, the_answer, f"{inc_rule}"))
    
    

def name_severity(inc_model: str,
                  inc_name:  str, 
                  inc_title: str, 
                  inc_severity:  str) -> str:
    
    the_answer=stig_prompt_builder("severity", inc_name, inc_title)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_severity}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_severity}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM, the_answer, f"{inc_severity}"))    

def name_desc(inc_model: str,
              inc_name:  str, 
              inc_title: str, 
              inc_desc:  str) -> str:
    
    the_answer=stig_prompt_builder("description", inc_name, inc_title)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_desc}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_desc}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_desc}"))    
    

###########################################
#- Fix
###########################################
def id_fix (inc_model: str,
            inc_name:  str, 
            inc_rule_id: str, 
            inc_fix:  str) -> str:

    the_answer=stig_prompt_builder("fix test", inc_name, inc_rule_id)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_fix}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_fix}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_fix}"))        


def name_fix(inc_model: str,
             inc_name:  str, 
             inc_stig_name: str, 
             inc_fix:  str) -> str:
    
    the_answer=stig_prompt_builder("description", inc_name, inc_stig_name)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_fix}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_fix}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_fix}"))        
    

###########################################
#- Check
###########################################
def id_check (inc_model: str,
              inc_name:  str, 
              inc_rule_id: str, 
              inc_check:  str) -> str:
    
    the_answer=stig_prompt_builder("check test", inc_name, inc_rule_id)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_check}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_check}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_check}"))        

    
def name_check(inc_model: str,
               inc_name:  str, 
               inc_stig_name: str, 
               inc_check:  str) -> str:
    
    the_answer=stig_prompt_builder("check test", inc_name, inc_stig_name)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_check}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_check}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_check}"))        

###########################################
#- Check & Fix 
###########################################

def fix_name_check_name(inc_model: str,
                        inc_name:  str, 
                        inc_fix_test:  str, 
                        inc_check_test: str) -> str:

    the_answer=stig_prompt_builder("check test", inc_name, inc_check_test)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_fix_test}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_fix_test}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_fix_test}"))        


    
def check_name_fix_name(inc_model: str,
                        inc_name:  str, 
                        inc_fix_test:  str, 
                        inc_check_test: str) -> str:

    the_answer=stig_prompt_builder("check test", inc_name, inc_check_test)
    if inc_model=="perf":
        return(create_perf_response_template(the_answer, f"{inc_check_test}"))
    elif inc_model=="pcpf":
        return(create_pcpf_response_template(the_answer, f"{inc_check_test}"))
    else:
        return(create_gpt35_response_template(PROMPT_PRE_SYSTEM,the_answer, f"{inc_check_test}"))        

```

# Iterate through each dataset and make the outputs

```python
def test_train_split(output_type:str,
                     save_location:str,
                     data_domain: [],
                     json_results: [],):

    print(f"......we are processing {output_type} data.")
    print("......" + str(type(data_domain)) + f" and size is: {len(data_domain)}")
    
    try:
        the_name=cleanse_text(json_results['Title'])
        stig_description=cleanse_text(json_results['Description'])
        stig_version=cleanse_text(json_results['Version'])
        stig_release=cleanse_text(json_results['Release'])
        stig_source=cleanse_text(json_results['Source'])

    except Exception as e:
        print(f"ERROR: In test_train_split() STIG details (title, description) pull from JSON issue: {str(e)}")
        return

    try:
        perf_filename=save_location+"perf"+"_"+output_type+".json"
        gpt35_filename=save_location+"gpt35"+"_"+output_type+".json"
        pcpf_filename=save_location+"pcpf"+"_"+output_type+".json"
        
        f_perf= open(perf_filename, "a")
        f_gpt35= open(gpt35_filename, "a")
        f_pcpf= open(pcpf_filename, "a")
        print(f"...saving output to: {perf_filename}, {pcpf_filename}, {gpt35_filename}")
    except Exception as e:
        print(f"ERROR In test_train_split() creating data filenames: {str(e)}, skipping output.")        
        return        

    try:
        for idx in range(0, len(data_domain)):
            try:
                the_rule=data_domain[idx]
                vuln_id=the_rule['VulnID']
                rule_id=the_rule['RuleID']
                stig_id=the_rule['StigID']
                severity=cleanse_text(the_rule['Severity'])
                cat=cleanse_text(the_rule['Cat'])
                classification=cleanse_text(the_rule['Classification'])
                title=cleanse_text(the_rule['RuleTitle'])
                description=the_rule['VulnDiscussion']
    
                check_test=""
                if "CheckText" in the_rule:
                    check_test=the_rule['CheckText']
                elif "check-content" in the_rule:
                    check_test=the_rule['check-content']
                else:
                    check_test="no check test"
    
                fix_test=""
                if "FixText" in the_rule:
                    fix_test=the_rule['FixText']
                elif "fixtext" in the_rule:
                    fix_test=the_rule['fixtext']
                else:
                    fix_test="no fix test"
            except Exception as e:
                print(f"ERROR In test_train_split() marshalling data: {str(e)}, skipping output.")        
                continue        
        
            try:
                f_perf.write(full_input("perf", the_name, rule_id, severity, title, description, fix_test, check_test) +"\n")
                f_pcpf.write(full_input("pcpf", the_name, rule_id, severity, title, description, fix_test, check_test) +"\n")
                f_gpt35.write(full_input("gpt35",the_name, rule_id, severity, title, description, fix_test, check_test) +"\n")
                
                f_perf.write(id_relation("perf",the_name, title, rule_id) +"\n")
                f_pcpf.write(id_relation("pcpf",the_name, title, rule_id) +"\n")
                f_gpt35.write(id_relation("gpt35",the_name, title, rule_id) +"\n")
                
                f_perf.write(name_severity("perf",the_name, title, severity) +"\n")
                f_pcpf.write(name_severity("pcpf",the_name, title, severity) +"\n")
                f_gpt35.write(name_severity("gpt35",the_name, title, severity) +"\n")
    
                
                f_perf.write(name_desc("perf",the_name, title, description) +"\n")
                f_pcpf.write(name_desc("pcpf",the_name, title, description) +"\n")
                f_gpt35.write(name_desc("gpt35",the_name, title, description) +"\n")
                  
                f_perf.write(id_fix("perf",the_name,rule_id, fix_test) +"\n")
                f_pcpf.write(id_fix("pcpf",the_name,rule_id, fix_test) +"\n")
                f_gpt35.write(id_fix("gpt35",the_name,rule_id, fix_test) +"\n")
                
                f_perf.write(name_fix("perf",the_name,rule_id, fix_test) +"\n")
                f_pcpf.write(name_fix("pcpf",the_name,rule_id, fix_test) +"\n")
                f_gpt35.write(name_fix("gpt35",the_name,rule_id, fix_test) +"\n")
            
                f_perf.write(name_check("perf",the_name,rule_id, check_test) +"\n")
                f_pcpf.write(name_check("pcpf",the_name,rule_id, check_test) +"\n")
                f_gpt35.write(name_check("gpt35",the_name,rule_id, check_test) +"\n")
    
                
                f_perf.write(fix_name_check_name("perf",the_name, fix_test, check_test) +"\n")
                f_pcpf.write(fix_name_check_name("pcpf",the_name, fix_test, check_test) +"\n")
                f_gpt35.write(fix_name_check_name("gpt35",the_name, fix_test, check_test) +"\n")
    
                
                f_perf.write(check_name_fix_name("perf",the_name, fix_test, check_test) +"\n")        
                f_pcpf.write(check_name_fix_name("pcpf",the_name, fix_test, check_test) +"\n")        
                f_gpt35.write(check_name_fix_name("gpt35",the_name, fix_test, check_test) +"\n")                    
                
            except Exception as e:
                print(f"ERROR In test_train_split() detected trying to write one of the output files as follows: {str(e)}, skipping output.")        
                return
    finally:
        f_perf.close()  
        f_pcpf.close()  
        f_gpt35.close()  
        print(f"...finished closing all files.")
```

```python
def generate_PromptInput(stig_input_filename: str,
                         save_location: str):

    tests=[]
    trains=[]
    the_rules=[]
    test_list={}
    
    try:
       
        json_results = convert_stig(stig_input_filename)
        print(f"...converting {stig_input_filename} to JSON")
    
        the_rules=json_results['Rules']
        test=int( int(len((the_rules)) * the_split))
        train=len(the_rules) - test
        print(f"...Train:{train}...Test:{test}...theRules:{len(the_rules)}")

        for idx in range(0, test):
            n = random.randint(0,len(the_rules))
#            #test_list.append(n)
            test_list[n]=1
#            #the_index=test_list[-1]
#            #tests.append(the_rules[the_index])
#            #del the_rules[the_index]
        the_idxs=list(test_list.keys())
        print(the_idxs)
        #tests.append(the_rules[test_list.keys()])
        tests = the_rules[the_idxs]
#        del the_rules[the_idxs]

        trains= the_rules[:]
        trains.append(tests)   #we're adding back in the tests to get a robust response.

        print(f"...Train[{len(trains)}]...Test[{len(tests)}]")
        
        print(f"...looping through {len(the_rules)} rules.")
        if len(the_rules) > 0:
            test_train_split("train", save_location, trains, json_results)
            test_train_split("test", save_location, tests, json_results)
        else:
            print(f"ERROR: zero results from the parsing of rules for {stig_input_filename}")
    except Exception as e:
        print(f"ERROR detected trying to process train/tests split in GeneratePrompt output as follows: {str(e)}, skipping output.")        
    finally:
        del trains
        del tests
        del the_rules
        del test_list
```

```python
import json
import tiktoken
import numpy as np
from collections import defaultdict

encoding = tiktoken.get_encoding("cl100k_base") # default encoding used by gpt-4, turbo, and text-embedding-ada-002 models

def num_tokens_from_messages(messages, tokens_per_message=3, tokens_per_name=1):
    num_tokens = 0
    for message in messages:
        num_tokens += tokens_per_message
        for key, value in message.items():
            num_tokens += len(encoding.encode(value))
            if key == "name":
                num_tokens += tokens_per_name
    num_tokens += 3
    return num_tokens

def num_assistant_tokens_from_messages(messages):
    num_tokens = 0
    for message in messages:
        if message["role"] == "assistant":
            num_tokens += len(encoding.encode(message["content"]))
    return num_tokens

def print_distribution(values, name):
    print(f"\n#### Distribution of {name}:")
    print(f"min / max: {min(values)}, {max(values)}")
    print(f"mean / median: {np.mean(values)}, {np.median(values)}")
    print(f"p5 / p95: {np.quantile(values, 0.1)}, {np.quantile(values, 0.9)}")

files = [SAVE_DIR+os.sep+'gpt35_training_set.jsonl', SAVE_DIR+os.sep+'gpt35_validation_set.jsonl']

for file in files:
    print(f"Processing file: {file}")
    with open(file, 'r', encoding='utf-8') as f:
        dataset = [json.loads(line) for line in f]

    total_tokens = []
    assistant_tokens = []

    for ex in dataset:
        messages = ex.get("messages", {})
        total_tokens.append(num_tokens_from_messages(messages))
        assistant_tokens.append(num_assistant_tokens_from_messages(messages))
    
    print_distribution(total_tokens, "total tokens")
    print_distribution(assistant_tokens, "assistant tokens")
    print('*' * 50)
```

from gpt_index import SimpleDirectoryReader, GPTListIndex, GPTSimpleVectorIndex, LLMPredictor, PromptHelper
from langchain import OpenAI
import gradio as gr
import sys
import os

os.environ["OPENAI_API_KEY"] = ''

def construct_index(directory_path):
    max_input_size = 4096
    num_outputs = 512
    max_chunk_overlap = 20
    chunk_size_limit = 600

    prompt_helper = PromptHelper(max_input_size, num_outputs, max_chunk_overlap, chunk_size_limit=chunk_size_limit)

    llm_predictor = LLMPredictor(llm=OpenAI(temperature=0.7, model_name="text-davinci-003", max_tokens=num_outputs))

    documents = SimpleDirectoryReader(directory_path).load_data()

    index = GPTSimpleVectorIndex(documents, llm_predictor=llm_predictor, prompt_helper=prompt_helper)

    index.save_to_disk('index.json')

    return index

def chatbot(input_text):
    index = GPTSimpleVectorIndex.load_from_disk('index.json')
    response = index.query(input_text, response_mode="compact")
    return response.response

iface = gr.Interface(fn=chatbot,
                     inputs=gr.inputs.Textbox(lines=7, label="Enter your text"),
                     outputs="text",
                     title="My AI Chatbot")

index = construct_index("docs")
iface.launch(share=True)

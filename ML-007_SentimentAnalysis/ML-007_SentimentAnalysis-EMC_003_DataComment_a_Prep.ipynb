```python
## Manages exception output.
#  @param   (Exception)             - Exception to expound upon
#  @returns (None)                  - None
def process_exception(inc_exception) -> None:
    print(f"{BOLD_START}(Exception encountered):{BOLD_END} {type(inc_exception).__name__}")
    print(f"Details: {str(inc_exception)}")
    print("Traceback:")
    traceback.print_exc()
```

```python
def profile_function(func):
    def wrapper(*args, **kwargs):
        pr = cProfile.Profile()
        pr.enable()
        result = func(*args, **kwargs)
        pr.disable()
        s = io.StringIO()
        sortby = SortKey.CUMULATIVE
        ps = pstats.Stats(pr, stream=s).sort_stats(sortby)
        ps.print_stats()
        print(s.getvalue())
        return result
    return wrapper
```

```python
## Load a dataframe from a data file
#
#  @param    (str)         - str             - Filename to read
#  @param    (str)         - str             - Delimiter to parse the data with
#  @returns (pd.DataFrame) - pd.DataFrame    - Actual DataFrame that encapsulates the data.
def load_dataframe(inc_filename:str, inc_encoding:str) -> pd.DataFrame:
    
    debug.msg_info(f"Entering {__name__} {inspect.stack()[0][3]}")
    
    inner_df=pd.DataFrame()
    if os.path.isfile(inc_filename):
      #open the file, read the contents and close the file
      try:
          inner_df = pd.read_csv(inc_filename, delimiter=f"{DELIM}")
      except Exception as e:
          process_exception(e)
          raise SystemExit("Stop right there!")  
    else:
      error_msg=f"File not found.  You attempted to read: {inc_filename}.  Verify its existence."
      debug.msg_error(error_msg)
      raise SystemExit("Stop right there!")
    
    if len(inner_df)<1:
        debug.msg_error("There is no content in your data variable.")
        debug.msg_debug("...Verify you copied the input file correctly.")
        raise SystemExit("Stop right there!")
        
    debug.msg_info(f"Exiting {__name__} {inspect.stack()[0][3]}")
    
    return inner_df
```

```python
## Main read routine of 001 output
#
#  @param (list)
def read_data(inc_years:list) -> None:

    debug.msg_info(f"Entering {__name__} {inspect.stack()[0][3]}")

    #establish data version, aligned with code, CARA_2020_coded.txt
    target_folder=DATA_DIR
    years=inc_years
    df_columns=["ProjectId","LetterId","CodedText"]
    df=pd.DataFrame()
    total_rows=0
    total_coded_rows=0
    
    for year in years:
        target_filename=f"{target_folder}" + os.sep + f"CARA_{str(year)}_coded.txt"
        try:
            current_df=load_dataframe(target_filename, ENCODING)
            pre_read_count=len(current_df)
            current_df.columns=df_columns
            current_df.dropna(subset=['CodedText'], inplace=True)
            post_read_count=len(current_df)
            
            total_rows += pre_read_count
            total_coded_rows += post_read_count
            
            debug.msg_debug(f"...Read: {pre_read_count:,}     Retained: {post_read_count:,}")
        except (pickle.UnpicklingError, FileNotFoundError, IOError, Exception)  as e:
            debug.msg_warning("FAILED to read the ASCII file, you might have corruption, investigate.")
            process_exception(e)

        try:
            save_results(year, current_df)
        except (FileNotFoundError, IOError, Exception)  as e:
            debug.msg_warning("FAILED to write the datafile, you might have corruption, investigate.")
            process_exception(e)
    
    debug.msg_debug(f"You have {total_rows:,} read in from the dataset.")    
    debug.msg_debug(f"You have {total_coded_rows:,} read in for Coded Comments to consider for analysis / validation.")
    
    debug.msg_info(f"Exited {__name__} {inspect.stack()[0][3]}")
```

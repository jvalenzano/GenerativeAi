````python
def profile_function(func):
    def wrapper(*args, **kwargs):
        pr = cProfile.Profile()
        pr.enable()
        result = func(*args, **kwargs)
        pr.disable()
        s = io.StringIO()
        sortby = SortKey.CUMULATIVE
        ps = pstats.Stats(pr, stream=s).sort_stats(sortby)
        ps.print_stats()
        print(s.getvalue())
        return result
    return wrapper
````

````python
## Read the string and clean up the markdown created by gemini
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @param (Text for type to transform) - str- String type
#  Hat tip to Yvan@google
def markdown_escaper(text: str, type: str = 'json'):
    escape_length = len(f'```{type}')
    if text[:escape_length] == f'```{type}' and text[-3:] == '```':
        return text[escape_length:-3]
    else:
        return text

## Read the markdown provided and turn it into JSON
#
#  @param (Text for text to clean) - str    - Markdown to transform
#  Hat tip to Yvan@google
def markdown_to_json(md: str):
    return json.loads(markdown_escaper(md))

## Read the contents of a text input and remove URL's, extra spaces, carriage returns, etc..
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
def clean_text(inc_str: str) -> str:
    resumeText = re.sub('httpS+s*', ' ', inc_str)  # remove URLs
    resumeText = re.sub('#S+', '', resumeText)  # remove hashtags
    resumeText = re.sub('@S+', '  ', resumeText)  # remove mentions
    resumeText = re.sub(r'\r', ' ', resumeText)
    resumeText = re.sub(r'\n', ' ', resumeText)
    resumeText = re.sub(' +', ' ', resumeText) # remove extra whitespace
    resumeText = re.sub(r'\t', ' ', resumeText) #remove tabs

    resumeText.rstrip()
    resumeText.lstrip()
    resumeText = re.sub('[%s]' % re.escape("""!"#$%&'()*+,-./:;<=>?@[]^_`{|}~"""), ' ', resumeText)  # remove punctuations
    resumtText = ''.join([i if ord(i) < 128 else ' ' for i in resumeText])

    resumeText=re.sub(r'\W+', ' ', resumeText)
    resumeText=re.sub(' +', ' ', resumeText)       
    # Remove punctuation
    #no_punctuation = (nopunc.translate(str.maketrans('', '', string.punctuation)) for nopunc in lower)
    #resumeText = ''.join(x for x in resumeText if x.isalnum())
    #resumeText = re.sub(r'[^x00-x7f]',r' ', resumeText) 
    #resumeText = re.sub('s+', ' ', resumeText)  # remove extra whitespace
    return resumeText

## Read the contents of a text input and remove URL's, extra spaces, carriage returns, etc..
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - All junk removed.
def clean_json(inc_str: str) -> str:
    resumeText = re.sub(r'\r', ' ', inc_str)
    resumeText = re.sub(r'\n', ' ', resumeText)
    resumeText = re.sub(' +', ' ', resumeText) # remove extra whitespace
    resumeText = re.sub(r'\t', ' ', resumeText) #remove tabs
    resumeText.rstrip()
    resumeText.lstrip()
    resumeText = re.sub('[%s]' % re.escape("""#*'"""), ' ', resumeText)  # remove punctuations
    return resumeText

## Read the contents of a text input remove stop words (nltk)
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - Stop words removed
def clean_stop_words(inc_str:str) -> str:
        filtered_list = []
        response=word_tokenize(inc_str)
        wordlist = [x for x in response if (len(x)>=2 and x.isalpha())]
        for word in wordlist:
            if word.casefold() not in stop_words:
              filtered_list.append(word)
        return str(' '.join(filtered_list))

   
## Read the contents of a text input and remove extra spaces
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - Extra spaces removed
def clean_string (inc_str:str) -> str:
        response=re.sub(r'\W+', ' ', inc_str)
        response=re.sub(' +', ' ', response)           
        return str(response)

## Read the contents of a text input modify string for stem words (nltk)
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - Stem words altered
def clean_stem_words(inc_str:str) -> str:
        filtered_list = []
        response=word_tokenize(inc_str)
        wordlist = [x for x in response if (len(x)>=2 and x.isalpha())]
        stemmed_words = [stemmer.stem(word) for word in wordlist]
        return str(' '.join(stemmed_words))

## Read the contents of a text input remove stop words (nltk)
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - Stop words removed
def clean_lemmatizer_words(inc_str:str) -> str:
        filtered_list = []
        response=word_tokenize(inc_str)
        wordlist = [x for x in response if (len(x)>=2 and x.isalpha())]
        lemmatized_words = [lemmatizer.lemmatize(word) for word in wordlist]
        return str(' '.join(lemmatized_words))

## Read the contents of a text cleanse the string
#
#  @param (Text for text to clean) - str    - Incoming text without modification.
#  @returns (Transformed text)     - str    - Complete purge of all content for NLP
def cleanse_string(inc_str: str) -> str:
    response=clean_string(inc_str)
    response=clean_text(response)
    response=clean_stop_words(response)
    response=clean_stem_words(response)
    response=clean_lemmatizer_words(response)
    response=word_tokenize(response)
    return response
````

````python
## Generic routine for reading any data, always come back with a pd.DataFrame
#
#. @param (str) - Data type to read [ASCII, PDFFORM, PDF, VTT, CSV]
#  @param (str) - Source directory
#  @param (str) - Source filename
def read_data(inc_type: str, inc_directory: str, inc_filename: str) -> pd.DataFrame:
    
    debug.msg_info(f"Entering {__name__} {inspect.stack()[0][3]}")
    target_filename=inc_directory + os.sep + inc_filename
    output_df = pd.DataFrame()
    try:
        #check for the file's existence
        if os.path.isfile(target_filename):
            if inc_type=="VTT":
                #####################################
                # - Voice Transcript
                #####################################   
                the_individuals={}
                the_conversation=[]
                for caption in webvtt.read(target_filename):
                    #print(caption.start)  # Start time
                    #print(caption.end)    # End time
                    #print(caption.text)   # Caption text
                    #print(caption.voice)   # Caption Voice
                    the_voice=clean_text(caption.voice)
                    the_words=clean_text(caption.text)
                    the_conversation.append(the_voice + " said " + the_words + ". ")
                    if not the_voice in the_individuals:
                        the_individuals[the_voice]=[]
                    the_individuals[the_voice].append(the_words + ". ")                    

                final_conversation=" ".join(the_conversation)
                output_df = pd.DataFrame({'Conversation': [str(final_conversation)]})
                
                for the_key in list(the_individuals.keys()):
                    #print(the_key)
                    my_series=pd.Series(" ".join(the_individuals[the_key]), name=f"{the_key}")
                    output_df[the_key]=my_series

            elif inc_type=="ASCII":
                #####################################
                # - Plain Text Read
                #####################################            
                with open(target_filename, "r") as the_file:
                    data=the_file.read()
                the_data = {'Data': data}
                output_df = pd.DataFrame(the_data)

                
        else:
            print("ERROR: File not found.  Check the previous code block to ensure you file copied.")
            print(f"...target file: {target_filename}")
            raise SystemError(f"Failed to find {target_filename}")
    except (Exception, IOError) as e:
        process_exception(e)
    
    debug.msg_info(f"Exiting {__name__} {inspect.stack()[0][3]}")
    return output_df
````

````python
# Validate requirements like file locations.
#
#  @param (None)
def validate() -> None:

    debug.msg_info(f"Entering {__name__} {inspect.stack()[0][3]}")

    try:
        if not (os.path.exists(DATA_DIR)):
            debug.msg_error(f"{DATA_DIR} does not appear to exist.")
            debug.msg_error("Check your paths, did you mount the proper file system?")
            raise SystemError
    except Exception as e:
        process_exception(e)
        raise SystemError
            
    debug.msg_info(f"Exited {__name__} {inspect.stack()[0][3]}")

````

````python

#
#  @param (None)
def process() -> None:

    debug.msg_info(f"Entering {__name__} {inspect.stack()[0][3]}")

    target_directory=OUTPUT_DIR + os.sep + f"{data_version_release}"
    try:
        if not os.path.isdir(target_directory):
            os.makedirs(target_directory)        
    except (Exception, IOError) as e:
        process_exception(e)
        raise SystemError
        
    ##########################################################
    #- Load data
    ##########################################################        
    debug.msg_debug("Loading data file(s).")

    day1_df=process_one_day(4, 1)
    if len(day1_df) > 0:
        debug.msg_debug("Day 1 successfully captured.")
        process_generative_results(day1_df, "day1")
    else:
        debug.msg_warning("Day 1 FAILED.")
    
    day2_df=process_one_day(4,2)    
    if len(day2_df) > 0:
        debug.msg_debug("Day 2 successfully captured.")
        process_generative_results(day2_df, "day2")
    else:
        debug.msg_warning("Day 1 FAILED.")
        
    day3_df=process_one_day(4,3)
    if len(day3_df) > 0:
        debug.msg_debug("Day 3 successfully captured.")
        process_generative_results(day3_df, "day3")
    else:
        debug.msg_warning("Day 3 FAILED.")    

    session1_df=process_one_session(1)
    if len(session1_df) > 0:
        debug.msg_debug("Session 1 successfully captured.")
        process_generative_results(session1_df, "session1")
    else:
        debug.msg_warning("Session 1 FAILED.")

    session2_df=process_one_session(2)
    if len(session2_df) > 0:
        debug.msg_debug("Session 2 successfully captured.")
        process_generative_results(session2_df, "session2")
    else:
        debug.msg_warning("Session 2 FAILED.")        
    
    session3_df=process_one_session(3)
    if len(session3_df) > 0:
        debug.msg_debug("Session 3 successfully captured.")
        process_generative_results(session3_df, "session3")
    else:
        debug.msg_warning("Session 3 FAILED.")        

    session4_df=process_one_session(4)
    if len(session4_df) > 0:
        debug.msg_debug("Session 4 successfully captured.")
        process_generative_results(session4_df, "session4")
    else:
        debug.msg_warning("Session 4 FAILED.")        
    
   
    
    debug.msg_info(f"Exited {__name__} {inspect.stack()[0][3]}")

````

````python
def process_one_day(inc_sessions: int, inc_value:int) -> pd.DataFrame:
    

    debug.msg_info(f"Entered {__name__} {inspect.stack()[0][3]}")    
    
    master_df= pd.DataFrame()
    if not isinstance(inc_value, int) and inc_value < 1:
        return master_df
    
    for idx, value in enumerate(range(1,inc_sessions+1)):
        target_directory=DATA_DIR +os.sep+ f"{data_version_release}" + os.sep + f"Session-{value}"
        debug.msg_debug(f"...access directory: {target_directory}")
        if os.path.isdir(target_directory):
            dir_list = os.listdir(target_directory)
            files = [f for f in dir_list if os.path.isfile(target_directory+ os.sep +f) and f.endswith(".vtt")]
            file=files[inc_value-1]
            debug.msg_debug(f"......processing file: {file}")
            inner_df=read_data("VTT", target_directory, file)    
            if len(inner_df) > 0:
                master_df = pd.concat([master_df, inner_df], axis=0)    # concatenating along rows    
        else:
            debug.msg_warning(f"...directory ({target_directory}) is invalid.")
    debug.msg_info(f"Exited {__name__} {inspect.stack()[0][3]}")
    
    return master_df
````

````python
def process_one_session(inc_value:int) -> pd.DataFrame:
    
    debug.msg_info(f"Entered {__name__} {inspect.stack()[0][3]}")    
    
    master_df= pd.DataFrame()
    if not isinstance(inc_value, int) and inc_value < 1:
        return master_df
    
    target_directory=DATA_DIR +os.sep+ f"{data_version_release}" + os.sep + f"Session-{inc_value}"
    dir_list = os.listdir(target_directory)
    files = [f for f in dir_list if os.path.isfile(target_directory+ os.sep +f)]
    for file in files:
        if file.endswith(".vtt") and not file.endswith("-Lab.vtt"):
            debug.msg_debug(f"...processing {file}")
            inner_df=read_data("VTT", target_directory, file)    
        if len(inner_df) > 0:
            master_df = pd.concat([master_df, inner_df], axis=0)    # concatenating along rows
        
    debug.msg_info(f"Exited {__name__} {inspect.stack()[0][3]}")    

    return master_df
````
